```python
import pandas as pd

df = pd.read_csv("../data/House_2.csv")

print(f"Kształt danych: {df.shape} \n")
print(f"Typy poszczególnych kolumn:\n{df.dtypes}")
```

```python

print(f"Zakres czasu: {df["Time"][0]} - {df["Time"].iloc[-1]}")
for q in [0.5, 0.9, 0.99, 0.999, 0.9999]:
    value = df["Aggregate"].quantile(q)
    count = (df["Aggregate"] > value).sum()
    count_p = round(100*count/df["Aggregate"].count(), 3)
    print(f"Per: {q} | value: {value}W | count: {count} ({count_p}%)")


print(f"Liczba próbek powyżej 15kW: {(df["Aggregate"] > 15_000).sum()}")
print(f"Liczba próbek powyżej 20kW: {(df["Aggregate"] > 20_000).sum()}")
```
